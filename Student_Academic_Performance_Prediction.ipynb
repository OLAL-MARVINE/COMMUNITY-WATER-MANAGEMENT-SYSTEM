```python

# ==========================================
# STUDENT ACADEMIC PERFORMANCE PREDICTION
# DATA SCIENCE PROJECT 2026/2027
# ==========================================

# Import libraries for data analysis
import pandas as pd
import numpy as np

# Import libraries for visualisation
import matplotlib.pyplot as plt
import seaborn as sns

# Import libraries for warnings and display
import warnings
warnings.filterwarnings('ignore')

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

print("All libraries imported successfully!")
```

```python

# Load the student dataset

file_path = "/content/Student Data set.xlsx"

# Read the Data worksheet
raw_data = pd.read_excel(
    file_path,
    sheet_name="Data"
)

# Read the Attributes worksheet
attributes = pd.read_excel(
    file_path,
    sheet_name="Attributes"
)

print("Dataset loaded successfully!")
print("Raw dataset shape:", raw_data.shape)
print("Attributes shape:", attributes.shape)
```

```python

# Display the first five records

display(raw_data.head())

# Display column names
print("\nDATASET COLUMN NAMES:")
print(raw_data.columns.tolist())

# Display dataset information
print("\nDATASET INFORMATION:")
raw_data.info()

# Display statistical summary
print("\nSTATISTICAL SUMMARY:")
display(raw_data.describe(include="all"))
```

```python

# Keep columns with actual variable names

valid_columns = [
    col for col in raw_data.columns
    if not str(col).startswith("Unnamed:")
]

df = raw_data[valid_columns].copy()

print("Working dataset created.")
print("Original dimensions:", raw_data.shape)
print("Working dataset dimensions:", df.shape)

print("\nWorking dataset columns:")
print(df.columns.tolist())
```

```python

# Check missing values

print("MISSING VALUES PER COLUMN")

missing_values = df.isnull().sum()

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage": (missing_values / len(df)) * 100
})

display(
    missing_summary[
        missing_summary["Missing Values"] > 0
    ].sort_values(
        by="Missing Values",
        ascending=False
    )
)

# Check duplicate records
print("\nDUPLICATE RECORDS:")
print(df.duplicated().sum())

# Check dataset dimensions
print("\nDATASET DIMENSIONS:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python

# ==========================================
# STEP 7: INVESTIGATE DUPLICATE RECORDS
# ==========================================

# Identify all records involved in duplication
duplicate_rows = df[df.duplicated(keep=False)]

print("Total records involved in duplication:",
      len(duplicate_rows))

# Display duplicate records
display(duplicate_rows)

# Show their original row indexes
print("\nOriginal row indexes:")
print(duplicate_rows.index.tolist())
```

```python

# Remove exact duplicate records

before_removal = len(df)

df = df.drop_duplicates(keep="first").copy()

after_removal = len(df)

print("Records before removal:", before_removal)
print("Records after removal:", after_removal)
print("Duplicates removed:", before_removal - after_removal)

# Confirm that no exact duplicates remain
print("Remaining duplicates:", df.duplicated().sum())
```

```python

# ==========================================
# STEP 8: TARGET VARIABLE INSPECTION
# ==========================================

print("KCSE GRADE FREQUENCY")

print(df["KCSEGrade"].value_counts(dropna=False))

print("\nKCSE GRADE PERCENTAGES")

print(
    df["KCSEGrade"]
    .value_counts(normalize=True, dropna=False)
    .mul(100)
    .round(2)
)

# Visualise KCSE grades
plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x="KCSEGrade",
    order=df["KCSEGrade"].value_counts().index
)

plt.title("Distribution of KCSE Grades")
plt.xlabel("KCSE Grade")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

```python

# ==========================================
# STEP 9: CATEGORICAL VALUE INSPECTION
# ==========================================

categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns

for column in categorical_columns:

    print("\nColumn:", column)

    print(
        df[column]
        .value_counts(dropna=False)
        .head(15)
    )
```

```python

# ==========================================
# STEP 10: STANDARDISE KCSE GRADES
# ==========================================

# Convert grades to uppercase and remove spaces
df["KCSEGrade"] = (
    df["KCSEGrade"]
    .astype("string")
    .str.strip()
    .str.upper()
)

# Display the corrected grade distribution
print("STANDARDISED KCSE GRADE DISTRIBUTION")

display(df["KCSEGrade"].value_counts(dropna=False))

print("\nUnique KCSE grades:")
print(df["KCSEGrade"].unique())
```

```python

# ==========================================
# STEP 11: CREATE CLASSIFICATION TARGET
# ==========================================

grade_mapping = {
    "A": "High",
    "B": "High",
    "C": "Average",
    "D": "Low",
    "E": "Low"
}

df["Performance"] = df["KCSEGrade"].map(grade_mapping)

print("STUDENT PERFORMANCE DISTRIBUTION")

display(df["Performance"].value_counts(dropna=False))

print("\nPerformance percentages:")

display(
    df["Performance"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

```python
nominal_features = [
    "Type_of_Institute",
    "County",
    "Gender",
    "Disability",
    "LivedP",
    "ConflictsP",
    "FStructure",
    "DFeesPay",
    "Sponsor",
    "CActivities",
    "Specialization",
    "Cschool",
    "Rclass",
    "F1Grade",
    "F2Grade",
    "F3Grade",
    "MockGrade",
    "ADrugs",
    "RModel",
    "Residence",
    "Scomposition",
    "Tlab",
    "Library",
    "Clab",
    "Electricity",
    "Internet"
]
```

```python
coded_features = [
    "Age",
    "PEmployed",
    "FEducation",
    "MEducation",
    "Cfrequency",
    "Fspecialization",
    "LStyle",
    "AStyle",
    "Tschool",
    "STlab",
    "SLibrary",
    "SClab",
    "SElectricity",
    "SInternet"
]
```

```python
# Combine the categorical feature groups
categorical_features = nominal_features + coded_features

# Check that every predictor has been classified
all_features = numerical_features + categorical_features

print("Numerical features:", len(numerical_features))
print("Nominal features:", len(nominal_features))
print("Coded categorical features:", len(coded_features))

print("\nTotal classified features:", len(all_features))
print("Total predictors in X:", X.shape[1])

# Check for missing or repeated feature names
print("\nUnclassified features:", set(X.columns) - set(all_features))
print("Unexpected features:", set(all_features) - set(X.columns))
print("Repeated feature names:", len(all_features) - len(set(all_features)))
```

```python
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

```python
# Create a modelling copy
X_model = X.copy()

# Convert categorical columns to object format
# Replace pandas missing values with NumPy NaN
X_model[categorical_features] = (
    X_model[categorical_features]
    .astype("object")
    .where(X_model[categorical_features].notna(), np.nan)
)

print("Modelling dataset shape:", X_model.shape)
print("Missing values:", X_model.isna().sum().sum())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_model,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training predictors:", X_train.shape)
print("Testing predictors:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
# Numerical preprocessing
numerical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical preprocessing
categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine both preprocessing pipelines
preprocessor = ColumnTransformer(transformers=[
    ("numerical", numerical_pipeline, numerical_features),
    ("categorical", categorical_pipeline, categorical_features)
])

print("Preprocessing pipeline created successfully.")
```

```python
from sklearn.dummy import DummyClassifier
from sklearn.pipeline import Pipeline
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# Create baseline model
baseline_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", DummyClassifier(strategy="most_frequent"))
])

# Train the baseline model
baseline_model.fit(X_train, y_train)

# Make predictions on test data
baseline_predictions = baseline_model.predict(X_test)

# Evaluate the baseline
print("BASELINE MODEL RESULTS")
print("----------------------")

print("Accuracy:",
      accuracy_score(y_test, baseline_predictions))

print("Balanced Accuracy:",
      balanced_accuracy_score(y_test, baseline_predictions))

print("Macro F1 Score:",
      f1_score(y_test, baseline_predictions, average="macro"))

print("\nClassification Report:")
print(classification_report(
    y_test,
    baseline_predictions,
    zero_division=0
))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, baseline_predictions))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)
```

```python
# Create Logistic Regression pipeline
logistic_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=42
    ))
])

# Train the model
logistic_model.fit(X_train, y_train)

# Predict performance on unseen test data
logistic_predictions = logistic_model.predict(X_test)
```

```python
print("LOGISTIC REGRESSION RESULTS")
print("---------------------------")

print("Accuracy:",
      accuracy_score(y_test, logistic_predictions))

print("Balanced Accuracy:",
      balanced_accuracy_score(y_test, logistic_predictions))

print("Macro F1 Score:",
      f1_score(y_test, logistic_predictions, average="macro"))

print("\nClassification Report:")
print(classification_report(
    y_test,
    logistic_predictions,
    zero_division=0
))

print("\nConfusion Matrix:")
print(confusion_matrix(
    y_test,
    logistic_predictions
))
```

```python
from sklearn.model_selection import (
    StratifiedKFold,
    GridSearchCV,
    cross_val_score
)
```

```python
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

models = {
    "Logistic Regression": logistic_model,
    "Decision Tree": decision_tree_model,
    "Random Forest": random_forest_model
}

for name, model in models.items():

    scores = cross_val_score(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring="f1_macro",
        n_jobs=-1
    )

    print("\n", name)
    print("Fold scores:", scores)
    print("Mean Macro F1:", scores.mean())
    print("Standard deviation:", scores.std())
```

```python
param_grid = {
    "classifier__n_estimators": [100, 200],
    "classifier__max_depth": [None, 10, 20],
    "classifier__min_samples_split": [2, 5],
    "classifier__min_samples_leaf": [1, 2, 4],
    "classifier__class_weight": [
        "balanced",
        "balanced_subsample"
    ]
}

grid_search = GridSearchCV(
    estimator=random_forest_model,
    param_grid=param_grid,
    scoring="f1_macro",
    cv=cv,
    n_jobs=-1,
    verbose=1
)

# Fit using training data only
grid_search.fit(X_train, y_train)

print("Best parameters:")
print(grid_search.best_params_)

print("\nBest cross-validation Macro F1:")
print(grid_search.best_score_)
```

```python
best_rf_model = grid_search.best_estimator_

tuned_predictions = best_rf_model.predict(X_test)

print("TUNED RANDOM FOREST RESULTS")
print("---------------------------")

print("Accuracy:",
      accuracy_score(y_test, tuned_predictions))

print("Balanced Accuracy:",
      balanced_accuracy_score(y_test, tuned_predictions))

print("Macro F1 Score:",
      f1_score(y_test, tuned_predictions, average="macro"))

print("\nClassification Report:")
print(classification_report(
    y_test,
    tuned_predictions,
    zero_division=0
))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, tuned_predictions))
```

```python
tuned_train_predictions = best_rf_model.predict(X_train)

print("Training Accuracy:",
      accuracy_score(y_train, tuned_train_predictions))

print("Testing Accuracy:",
      accuracy_score(y_test, tuned_predictions))
```

```python
from imblearn.over_sampling import RandomOverSampler
from imblearn.pipeline import Pipeline as ImbPipeline
from sklearn.ensemble import RandomForestClassifier
```

```python
oversampling_model = ImbPipeline(steps=[
    ("preprocessor", preprocessor),

    ("oversampler", RandomOverSampler(
        sampling_strategy="not majority",
        random_state=42
    )),

    ("classifier", RandomForestClassifier(
        n_estimators=200,
        max_depth=20,
        min_samples_leaf=4,
        min_samples_split=2,
        class_weight=None,
        random_state=42,
        n_jobs=-1
    ))
])
```

```python
oversampling_model.fit(X_train, y_train)

oversampling_predictions = oversampling_model.predict(X_test)
```

```python
print("RANDOM OVERSAMPLING RESULTS")
print("---------------------------")

print("Accuracy:",
      accuracy_score(y_test, oversampling_predictions))

print("Balanced Accuracy:",
      balanced_accuracy_score(y_test, oversampling_predictions))

print("Macro F1 Score:",
      f1_score(y_test, oversampling_predictions, average="macro"))

print("\nClassification Report:")
print(classification_report(
    y_test,
    oversampling_predictions,
    zero_division=0
))

print("\nConfusion Matrix:")
print(confusion_matrix(
    y_test,
    oversampling_predictions
))
```

```python
oversampling_cv_scores = cross_val_score(
    oversampling_model,
    X_train,
    y_train,
    cv=cv,
    scoring="f1_macro",
    n_jobs=-1
)

print("RANDOM OVERSAMPLING CROSS-VALIDATION")
print("------------------------------------")
print("Fold scores:", oversampling_cv_scores)
print("Mean Macro F1:", oversampling_cv_scores.mean())
print("Standard Deviation:", oversampling_cv_scores.std())
```

```python
# Predictions on training data
oversampling_train_predictions = oversampling_model.predict(X_train)

# Calculate accuracy
oversampling_train_accuracy = accuracy_score(
    y_train,
    oversampling_train_predictions
)

oversampling_test_accuracy = accuracy_score(
    y_test,
    oversampling_predictions
)

print("OVERSAMPLING MODEL: OVERFITTING CHECK")
print("-------------------------------------")
print("Training Accuracy:", oversampling_train_accuracy)
print("Testing Accuracy:", oversampling_test_accuracy)
print(
    "Accuracy Gap:",
    oversampling_train_accuracy - oversampling_test_accuracy
)
```
